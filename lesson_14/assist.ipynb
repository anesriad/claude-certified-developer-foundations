Goal:

> Prepare a concise client-call prep note. Check the current time, keep the final note under 25 words, verify the word count, revise if needed, then save the final note.

1. `get_time`
2. draft a note
3. `count_words`
4. revise if needed
5. `count_words` again
6. `save_note`
7. final answer
